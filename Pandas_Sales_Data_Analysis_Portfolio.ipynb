# Sales Data Analysis Using Pandas

## Project Overview
This project demonstrates how Python and Pandas can be used to explore, analyse and visualise a small business sales dataset.

The analysis examines monthly sales, order volumes and customer satisfaction to identify trends and useful business insights.

### Tools Used
- Python
- Pandas
- Matplotlib
- Jupyter Notebook

## 1. Import Libraries

```python
import pandas as pd
import matplotlib.pyplot as plt
```

## 2. Create and Display the Dataset

```python
data = {
    "Month": ["January", "February", "March", "April", "May", "June",
              "July", "August", "September", "October", "November", "December"],
    "Sales": [12500, 13800, 14200, 15100, 16500, 17200,
              18000, 17600, 18900, 20100, 21500, 22800],
    "Orders": [125, 138, 142, 151, 165, 172,
               180, 176, 189, 201, 215, 228],
    "Customer_Satisfaction": [82, 84, 83, 86, 87, 88,
                              89, 88, 90, 91, 92, 93]
}

df = pd.DataFrame(data)
df
```

## 3. Basic Data Exploration

```python
print("Dataset shape:", df.shape)
print("\nSummary statistics:")
display(df.describe())
```

## 4. Sales Analysis

Calculate total annual sales, average monthly sales and the month with the highest sales.

```python
total_sales = df["Sales"].sum()
average_sales = df["Sales"].mean()
highest_sales = df["Sales"].max()
highest_sales_month = df.loc[df["Sales"].idxmax(), "Month"]

print(f"Total annual sales: R{total_sales:,.2f}")
print(f"Average monthly sales: R{average_sales:,.2f}")
print(f"Highest sales: R{highest_sales:,.2f} in {highest_sales_month}")
```

## 5. Customer Satisfaction Analysis

```python
average_satisfaction = df["Customer_Satisfaction"].mean()
print(f"Average customer satisfaction: {average_satisfaction:.1f}%")
```

## 6. Visualisation

The line chart below shows how monthly sales changed throughout the year.

```python
plt.figure(figsize=(12, 5))
plt.plot(df["Month"], df["Sales"], marker="o", label="Monthly Sales")
plt.title("Monthly Sales Trend")
plt.xlabel("Month")
plt.ylabel("Sales (R)")
plt.xticks(rotation=45)
plt.grid(True, alpha=0.3)
plt.legend()
plt.tight_layout()
plt.show()
```

## 7. Key Findings

- Total annual sales were **R206,200.00**.
- Average monthly sales were approximately **R17,183.33**.
- The highest monthly sales were recorded in **December**, at **R22,800.00**.
- Average customer satisfaction was **87.75%**.
- Overall, the data shows a generally increasing sales trend during the year.

## 8. Conclusion

The analysis demonstrates how Pandas can be used to organise and analyse business data efficiently. The visualisation makes the sales trend easier to understand and supports the numerical findings.
